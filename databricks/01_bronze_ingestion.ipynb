```python
orders_df = (
    spark.read
    .option("header", True)
    .csv("/Volumes/workspace/retail/raw_files/orders.csv")
)

display(orders_df)
```

```python
orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.retail.bronze_orders")
```

```python
tables = [
    "categories",
    "products",
    "stores",
    "customers",
    "order_items",
]

for table in tables:
    df = (
        spark.read
        .option("header", True)
        .csv(f"/Volumes/workspace/retail/raw_files/{table}.csv")
    )

    target = f"workspace.retail.bronze_{table}"

    df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(target)

    print(f"{target}: {spark.table(target).count()} rows")
```
